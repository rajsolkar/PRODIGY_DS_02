# Handling Outliers
How are we going to identify outliers
- Z-score: here we measure how far the data point from the mean in term of standard deviation. If the value is +-3 SD from the mean then we consider it as outlier
- Interquartile Range (IQR): 
    <br>IQR = Q3 - Q1
    <br>Lower Bound = Q1 - 1.5 * IQR
    <br>Upper Bound = Q3 + 1.5 * IQR
    <br>Any data point that falls below the lower bound or above the upper bound is considered an outlier
- Boxplot: graphical representation of the data that highlights median, quartiles, upper and lower whisker. any value that lies outside the whiskers is considered an outlier

# Handling Outliers

```python
for make in df['make'].unique():
    price=df[df['make']==make]['price']
    q1=np.quantile(price,0.25)
    q3=np.quantile(price,0.75)
    iqr=q3-q1
    uw=q3+(1.5*iqr)
    lw=q1-(1.5*iqr)
    index_above_uw=price[price>uw].index
    index_below_lw=price[price<lw].index
    #print(index_above_uw)
    #print(index_below_lw)
    if len(index_above_uw) !=0:
        df.drop(index_above_uw,inplace=True)
    if len(index_below_lw) !=0:
        df.drop(index_below_lw,inplace=True)
        
    
```

##### One Hot Encoding

- This encoding will create a binary column for each unique category in the categorical feature where each column will contain 0s and 1s to indicate the presence of the category.
- It is preferable when the column contain few unique categories
